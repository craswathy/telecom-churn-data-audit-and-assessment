```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print("Nulls found after fixing type:", df['TotalCharges'].isnull().sum())
```

```python
df['TotalCharges'] = df['TotalCharges'].fillna(0)
print("Nulls after fill:", df['TotalCharges'].isnull().sum())
```

```python
print("Duplicates before:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after:", df.duplicated().sum())
```

```python
text_cols = df.select_dtypes(include='object').columns.drop('customerID')
for col in text_cols:
    print(col, ":", df[col].unique())
```

```python
for col in ['tenure', 'MonthlyCharges', 'TotalCharges']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    low = Q1 - 1.5 * IQR
    high = Q3 + 1.5 * IQR
    outliers = df[(df[col] < low) | (df[col] > high)]
    print(col, "-> outliers found:", len(outliers))
```

```python
df['customerID'] = df['customerID'].astype(str)
df['TotalCharges'] = df['TotalCharges'].astype(float)
df['MonthlyCharges'] = df['MonthlyCharges'].astype(float)
df['tenure'] = df['tenure'].astype(int)
print(df.dtypes)
```

```python
summary = pd.DataFrame({
    'Check': ['Rows', 'Nulls (isnull)', 'Duplicates', 'TotalCharges dtype'],
    'Before': [7043, 0, 0, 'text (object)'],
    'After': [df.shape[0], df.isnull().sum().sum(), df.duplicated().sum(), df['TotalCharges'].dtype]
})
print(summary)
```

```python
df.to_csv('telco_churn_cleaned.csv', index=False)
print("Saved. Final shape:", df.shape)
```
